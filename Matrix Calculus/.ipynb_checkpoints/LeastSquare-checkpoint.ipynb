```python
import numpy as np

# =====================================================================
# 1. Prepare Complex Dataset (A is a 3x2 Matrix)
# =====================================================================
# Columns representing: [Size, Number of Rooms]
A = np.array([[1.0, 1.0],
              [2.0, 1.0],
              [1.0, 2.0]])

# True prices vector b (shape: 3,)
b = np.array([6.0, 9.0, 8.0])

print("--- Multi-Feature Dataset Prepared ---")
print(f"Feature Matrix A (3x2):\n{A}")
print(f"Target Vector b (3x1): {b}\n")
```

```python
# =====================================================================
# 2. Run Normal Equation: x = (A.T @ A)^-1 @ A.T @ b
# =====================================================================
ATA = A.T @ A
ATA_inv = np.linalg.inv(ATA)
x = ATA_inv @ A.T @ b

print("=== [Result] Multi-Dimensional Normal Equation ===")
print(f"Optimal Parameter Vector x:\n{x}")
```

```python
# =====================================================================
# 3. Compute Predictions and Evaluate Euclidean Norm Loss Squared
# =====================================================================
# Predictions (Ax) will be a vector of shape (3,)
predictions = A @ x
errors = predictions - b

# Total Loss ||Ax - b||_2^2
loss_squared = errors @ errors

print("=== [Verification] Performance Evaluation ===")
print(f"Model Predictions (Ax): {predictions}")
print(f"Residual Errors (Ax - b): {errors}")
print(f"Minimum Possible Loss (Euclidean Norm Squared): {loss_squared:.4f}")
print("Conclusion: The system smoothly solved multiple equations simultaneously!")
```
