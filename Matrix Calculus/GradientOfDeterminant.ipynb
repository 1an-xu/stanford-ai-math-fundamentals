```python
import numpy as np

# To ensure the matrix is symmetric positive-definite (as required in Section 4.5), 
# we generate a random matrix and construct a positive-definite matrix via B = A.T @ A.
np.random.seed(42)
seed_matrix = np.random.rand(3, 3)
A = seed_matrix.T @ seed_matrix
print(A)
```

```python
# ==========================================
# 1. Verifying ∇|A| = |A| * A^{-T}
# ==========================================
det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
grad_det_analytical = det_A * inv_A.T
h = 1e-6  # A tiny step size for numerical differentiation

# Numerical solution (Brute-force calculation using tiny changes in the determinant)
grad_det_numerical = np.zeros_like(A)
for i in range(A.shape[0]):
 for j in range(A.shape[1]):
     A_plus = A.copy()
     A_minus = A.copy()
     A_plus[i,j] += h
     A_minus[i, j] -= h
     grad_det_numerical[i, j] = (np.linalg.det(A_plus) - np.linalg.det(A_minus)) / (2 * h)

print("1. Determinant Gradient Verification:")
print("Analytical Gradient:\n", grad_det_analytical)
print("Numerical Gradient:\n", grad_det_numerical)
print("Are they extremely close?:", np.allclose(grad_det_analytical, grad_det_numerical, atol=1e-5))
print("-" * 50)
```

```python
# ==========================================
# 2. Verifying ∇ log|A| = A^{-1}
# ==========================================
# Analytical solution
grad_logdet_analytical = inv_A  # The transpose is dropped due to symmetry

# Numerical solution (Brute-force calculation using tiny changes in the determinant)
grad_logdet_numerical = np.zeros_like(A)
for i in range(A.shape[0]):
 for j in range(A.shape[1]):
     A_plus = A.copy()
     A_minus = A.copy()
     A_plus[i,j] += h
     A_minus[i, j] -= h
     grad_logdet_numerical[i, j] = (np.log(np.linalg.det(A_plus)) - np.log(np.linalg.det(A_minus))) / (2 * h)

print("2. log|A| Gradient Verification:")
print("Analytical Gradient:\n", grad_logdet_analytical)
print("Numerical Gradient:\n", grad_logdet_numerical)
print("Are they extremely close?:", np.allclose(grad_logdet_analytical, grad_logdet_numerical, atol=1e-5))
```
