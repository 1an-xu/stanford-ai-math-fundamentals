```python
def function_taker(a, func):
    return func(a)

def parameter_function(x):
    return x + 5

print(function_taker(1.0, int))
print(function_taker(1.0, parameter_function))
num_to_add = 10
print(function_taker(1.0, lambda x: x + num_to_add))
```

```python
class Foo():
    class_variable = 'same'

    def __init__(self, x):
        self.x = x

    def print_x(self):
        print(self.x)

    @classmethod
    def modify_class_variable(cls):
        cls.class_variable = 'changed'

    @staticmethod
    def print_hello():
        print('Hello')

print(Foo.class_variable)
obj1 = Foo(6)
obj1.print_x()
print(obj1.class_variable)

obj2 = Foo(5)
obj2.print_x()
print(obj2.class_variable)

obj1.modify_class_variable()
print(obj1.class_variable)
print(obj2.class_variable)
print(Foo.class_variable)
```

```python
adict = {'dog': 10, 'bird': 5, 'lion': 8}
for key in adict:
    print(key, adict[key])

for key, value in adict.items():
    print(key, value)

from collections import defaultdict

adict = defaultdict(int)
adict['cat'] = 1
print(adict['cat'])
print(adict['dog'])

adict = defaultdict(lambda: 'unknown')
adict['cat'] = 'feline'
print(adict['cat'])
print(adict['dog'])
```

```python
from collections import Counter
counter1 = Counter()
counter1['t'] = 10
counter1['t'] += 1
counter1['e'] += 1
print(counter1)

counter2 = Counter("letters to be counted")
print(counter2)

print("1", counter1 + counter2) #merge counts
print("2", counter1 - counter2)
print("3", counter1 or counter2) # or for intersection 
print("4", counter1 and counter2) # and for union

for k, v in counter2.most_common(5):
    print(k, v)
```
