```python
import torch
# permutation of tensors
x = torch.rand((2, 3))
y = torch.einsum("ij -> ji", x) #transpose
print(x)
print(y)
```

```python
# Summation
y = torch.einsum("ij ->", x)
print(y)

# Column sum
y = torch.einsum("ij -> j", x)
print(y)

# Row sum
y = torch.einsum("ij -> i", x)
print(y)
```

```python
# matrix multiplication
x = torch.rand((2, 3))
v = torch.rand((1, 3))
# x @ v.T
y = torch.einsum("ij, kj -> ik", x, v)
print(y)
# x @ x.T
y = torch.einsum("ij, kj -> ik", x, x)
print(y)
```

```python
# Dot product first row with first row of matrix
y = torch.einsum("i, i ->", x[0], x[0])
print(y)

# Dot product with matrix
x = torch.ones((2, 3))
y = torch.einsum("ij, ij ->", x, x * 2)
print(x)
print(y)

# Hadamard product (element wise multiplication)
y = torch.einsum("ij, ij -> ij", x, x * 2)
print(y)
```

```python
# outer product
a = torch.ones((3)) * 2
b = torch.ones((5)) * 3
y = torch.einsum("i, j -> ij", a, b)
print(y)

# batch matrix multiplication
a = torch.ones((3, 2, 5))
b = torch.ones((3, 5, 3))
y = torch.einsum("bij, bjk -> bik", a, b)
print(y)
```

```python
# matrix diagonal
x = torch.rand((3, 3))
y = torch.einsum("ii -> i", x)
print(x)
print(y)

# matrix trace
y = torch.einsum("ii ->", x)
print(y)
```
